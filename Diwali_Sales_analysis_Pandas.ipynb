**Library Import**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

**Load data**

```python
df = pd.read_csv("/content/sample_data/Diwali Sales Data.csv", encoding="unicode_escape")
```

**Inspect Data**

```python
df.shape
```

```python
df.info()
```

```python
df.head(10)
```

```python
df.tail(5)
```

**Cleaning and Preprocessing**

```python
#Drop unrelated and blank Col
df.drop(["Status","unnamed1"],axis=1, inplace=True)
```

```python
df.shape
```

```python
#Duplicates
df.duplicated().sum()
```

```python
df.drop_duplicates()
```

```python
#check for Nullvalues
df.isnull().sum()
```

```python
#Drop Null Values
df.dropna(inplace=True)
```

```python
df.dtypes
```

```python
#change amount to int
df["Amount"] = df["Amount"].astype("int")
df["Amount"].dtype
```

```python
#stats for each column
df.describe()
```

```python
df.columns
```

**EDA(Exploratory Data Analysis)**

Gender

```python
count_gender = sns.countplot(x="Gender",data=df)
for i in count_gender.containers:
  count_gender.bar_label(i)
```

```python
#Gender wise Sales
gender_sales = df.groupby(["Gender"],as_index=False)["Amount"].sum()
print(gender_sales)
```

```python
sns.barplot(x="Gender",y="Amount",data=gender_sales)
```

*Insight : From above graphs we can see that most of the buyers are females and even the purchasing power of females are greater than men*

**AGE group**

```python
agegroup_count=sns.countplot(x="Age Group",data=df,hue="Gender")
for i in agegroup_count.containers:
  agegroup_count.bar_label(i)
```

```python
#Amount by Age Group
df.groupby(["Age Group"])["Amount"].sum().sort_values(ascending=False).plot(kind="bar")
```

*From above graphs we can see that most of the buyers are of age group between 26-35 yrs female*

**State**

```python
#total number of orders from top 10 States
state_by_order=df.groupby(["State"],as_index=False)["Orders"].sum().sort_values(by="Orders",ascending=False).head(10)
print(state_by_order)
```

```python
sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(x="State",y="Orders",data=state_by_order)
```

*Insight from above charts is UP ,Maharshtra , Karnataka and delhi are the top state by Orders*

```python
state_by_amount=df.groupby(["State"], as_index=False)["Amount"].sum().sort_values(by="Amount",ascending=False).head(10)

sns.barplot(y="Amount",x="State",data= state_by_amount)
```

*Insight from above charts is UP ,Maharshtra , Karnataka and delhi are the top state by Amount*

**Marital_Status**

```python
count_marital = sns.countplot(x="Marital_Status",data=df,hue="Gender" )


for i in count_marital.containers:
  count_marital.bar_label(i)
```

```python
# Marital and gender by amount
marital_by_amount=df.groupby(["Marital_Status","Gender"],as_index=False)["Amount"].sum().sort_values(by="Amount",ascending=False)
print(marital_by_amount)
```

```python
sns.set(rc={"figure.figsize":(6,5)})
sns.barplot(y="Amount",x="Marital_Status",data=marital_by_amount,hue="Gender")
```

*From above graphs we can see that most of the buyers are married (women) and they have high purchasing power*

**Occupation**

```python
sns.set(rc={"figure.figsize":(22,5)})
a=sns.countplot(x="Occupation",data=df)

for i in a.containers:
  a.bar_label(i)
```

```python
#occupation by amount
occ_by_amt = df.groupby(["Occupation"],as_index=False)["Amount"].sum().sort_values(by="Amount",ascending=False).head(10)
print(occ_by_amt)
```

```python
sns.barplot(x="Occupation",y="Amount",data=occ_by_amt)
```

*Insight : From the above Graph IT sector , Healthcare and aviation occupation type spends more money*

**Product Category**

```python
sns.set(rc={"figure.figsize":(30,10)})
sns.countplot(x="Product_Category",data=df)
```

```python
# product category by amount
pro_amt = df.groupby(["Product_Category"],as_index=False)["Amount"].sum().sort_values(by="Amount",ascending=False).head(10)
print(pro_amt)
```

```python
sns.barplot(x="Product_Category",y="Amount",data=pro_amt)
```

*Insight : From the above most of the sold product are Food , clothing , electronics and gadgets by amount*

**Orders by ProudctID**

```python
#order by Product id
df.groupby("Product_ID")["Orders"].sum().nlargest(10).sort_values(ascending=False).plot(kind="bar")
```

*Insight : 10 highest orders by the product_id from above Graph*

# **CONCLUSION**

*Married women age group 26-35 yrs from UP, Maharastra and Karnataka working in IT, Healthcare and Aviation are more likely to buy products from Food. Clothing and Electronics category*
